```python
# not important in my project
columns=[
        'id',
        'imdb_id',
        'popularity',
        'budget',
        'revenue',
        'homepage',
        'tagline',
        'production_companies',
        'vote_count',
        'budget_adj',
        'revenue_adj'
 ]
movies.drop(
    columns,axis=1,
    inplace=True
)
```

```python
movies.columns
```

```python
movies.isnull().sum()
```

```python
movies['genres']=movies['genres'].fillna('Unknown')
movies['cast']=movies['cast'].fillna('Unknown')
movies['director']=movies['director'].fillna('Unknown')
movies['keywords']=movies['keywords'].fillna('no Keywords')
```

```python
movies.isnull().sum()
```

```python
movies.dropna(subset=['overview'], inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.drop_duplicates(inplace=True)
```

```python
movies.reset_index(drop=True, inplace=True)
```

```python
movies.to_csv("clean_movies.csv", index=False)
```
