# **making document for every movie**  

```python
import pandas as pd
```

```python
movies=pd.read_csv('cleanmovies.csv')
def create_movie_text(filed):
  text=f"""
  Title : {filed['original_title']}

  Overview:
    {filed['overview']}
   Genres:
    {filed['genres']}

   Director:
    {filed['director']}

  Cast:
    {filed['cast']}

  Keywords:
    {filed['keywords']}

  Runtime:
    {filed['runtime']} minutes

  Release Year:
    {filed['release_year']}

  Rating:
    {filed['vote_average']}
    """

  return text

```

```python
movies['text']=movies.apply(create_movie_text,axis=1)
```

```python
print(movies['text'][0])
```

```python
movies.to_csv('movies_doc.csv',index=
              False)
```
